# Chargement de donnees

```python
import pandas as pd

# Charger les données
df = pd.read_csv("../data/superstore.csv")

# Aperçu
print(df.head())
```

# Structure du dataset

```python
df.info()
```

# Nettoyage 

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

# Valeurs manquantes

```python
df.isnull().sum()
```

# Postal Code contient 11 valuers manquantes dont il faudra supprimer

```python
df['postal_code'] = df['postal_code'].fillna(0)
```

# Supprimer les doublons

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

# Conversion des donnees

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='%d/%m/%Y')
```

# Feature engineering

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
```

# Verification finale

```python
df.describe()
```

# Sauvegarder les donnees propres

```python
df.to_csv("../data/cleaned_superstore.csv", index=False)
```
