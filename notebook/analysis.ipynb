# Chargement de donnees

```python
import pandas as pd

# Charger les données
df = pd.read_csv("../data/superstore.csv")

# Aperçu
print(df.head())
```

# Structure du dataset

```python
df.info()
```

# Nettoyage 

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

# Valeurs manquantes

```python
df.isnull().sum()
```

# Postal Code contient 11 valuers manquantes dont il faudra supprimer

```python
df['postal_code'] = df['postal_code'].fillna(0)
```

# Supprimer les doublons

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

# Conversion des donnees

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='%d/%m/%Y')
```

# Feature engineering

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
```

# Verification finale

```python
df.describe()
```

# Sauvegarder les donnees propres

```python
df.to_csv("../data/cleaned_superstore.csv", index=False)
```

# ===========================
# KEY PERFORMANCE INDICATORS
# ===========================

```python
print(df.columns)
```

# Générer un profit entre -20% et +30% des ventes

```python
import numpy as np

df['profit'] = df['sales'] * np.random.uniform(-0.2, 0.3, size=len(df))
print(df.profit)
```

```python
df.to_csv("../data/cleaned_superstore.csv", index=False)
```

# Compute total sales, profits, and orders

```python
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_orders = df['order_id'].nunique()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
```

#  Insight: 
# The company genreated X revenue with Y profit across Z orders.
# indication a heathy business performance.

# Top produits

```python
top_products = df.groupby('product_name')['sales'].sum() \
                 .sort_values(ascending=False).head(10)

top_products
```

# Insight:
# A small number of products generate a large portion of revenue (Pareto effect).

# Categories les plus rentables

```python
category_profit = df.groupby('category')['profit'].sum() \
                    .sort_values(ascending=False)

category_profit
```

# Insight:
# Technology category is the most profitable, while Furniture shows lower margins.

# Analyse temporelle

```python
monthly_sales = df.groupby(['year', 'month'])['sales'].sum().reset_index()

monthly_sales
```

# Ventes par mois 

```python
df.groupby('month_name')['sales'].sum().sort_values(ascending=False)
```

# Insight:
# Sales peak during end-of-year months, likely due to seasonal demand.

# Performance par region

```python
region_sales = df.groupby('region')['sales'].sum().sort_values(ascending=False)
region_profit = df.groupby('region')['profit'].sum().sort_values(ascending=False)

region_sales
region_profit
```

# Insight:
# Some regions generate high sales but low profit → potential inefficiency.

# Top clients

```python
top_customers = df.groupby('customer_name')['sales'].sum() \
                  .sort_values(ascending=False).head(10)

top_customers
```

# Insight:
# A small number of customers contribute significantly to total revenue.

# Detection des problemes business (tres fort en entretien)

```python
loss_making = df[df['profit'] < 0]
loss_making.groupby('category')['profit'].sum()
```

# Insight:
# Certain categories generate consistent losses and should be investigated.
